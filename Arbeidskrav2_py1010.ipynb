```python
import numpy as np

a = float(input("Skriv inn lengden på kortsiden av trekanten:"))

b = float(input("Skriv inn høyden på trekanten:"))

#Areal halvsirkel
r = float(a)/2
areal_halvsirkel = (np.pi * r **2)/2

#Omkrets buen til halvsirkelen
bue = np.pi * r

#Areal trekant
areal_trekant = (a*b)/2

#Hypotenus
hyp = np.sqrt(a**2 + b**2)

#Totalt areal
areal_total = areal_halvsirkel + areal_trekant

print("Totalt areal er", round(areal_total, 3))

#Total omkrets

print ("Total omkrets er", round(b+hyp+bue, 3))

```
